```python
from bert_score import BERTScorer

scorer = BERTScorer(
    model_type='microsoft/deberta-v2-xxlarge-mnli',
    num_layers=40
)
scorer._tokenizer.model_max_length = 512

pairs = [
    ("Good", "Bad"),
    ("High", "Low"),
    ("Good", "The applicant has a strong credit history with no defaults."),
    ("Good", "Excellent"),
    ("Good", "xyz123"),
]

for ref, cand in pairs:
    P, R, F1 = scorer.score([cand], [ref])
    print(f"ref='{ref}' | cand='{cand[:40]}' | F1={F1.mean():.4f}")
```
