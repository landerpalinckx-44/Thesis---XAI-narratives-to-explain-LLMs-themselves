```python
from bert_score import BERTScorer

scorer = BERTScorer(model_type='roberta-large')

pairs = [
    ("Good", "Bad"),        # close in embedding space, but opposite meaning
    ("High", "Low"),        # same claim for stroke experiment
    ("Good", "The applicant has a strong credit history with no defaults."),  # clearly different
    ("Good", "Excellent"),  # true synonym as upper bound
    ("Good", "xyz123"),     # nonsense as lower bound
]
for ref, cand in pairs:
    P, R, F1 = scorer.score([cand], [ref])
    print(f"ref='{ref}' | cand='{cand[:40]}' | F1={F1.mean():.4f}")
```
